```python
import boto3, os

s3_client = boto3.client('s3')
os.makedirs("data", exist_ok=True)

response = s3_client.list_objects_v2(Bucket="dsan6000-wikipedia", Prefix="hourly_parquet/")
for obj in response['Contents']:
    key = obj['Key']
    filename = os.path.basename(key)
    s3_client.download_file("dsan6000-wikipedia", key, f"data/{filename}")
    print(f"Downloaded {filename}")
```
